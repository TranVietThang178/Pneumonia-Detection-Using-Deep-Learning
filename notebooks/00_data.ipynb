## 1. Imports

````python
import os
import zipfile
import shutil
````

## 2. Setup Output Directory

The raw data directory is cleared before downloading to ensure a clean state.

````python
data_dir = os.path.join('..', 'data', 'raw')

shutil.rmtree(data_dir, ignore_errors=True)
os.makedirs(data_dir, exist_ok=True)

print(f"Data directory ready: {os.path.abspath(data_dir)}")
````

## 3. Download Dataset from Kaggle

The dataset is downloaded using the Kaggle CLI included in `requirements.txt`.
If the download fails, try upgrading the library first:

```bash
pip install --upgrade kaggle
```

Alternatively, download the zip manually from Kaggle and place it in `data/raw/`.

````python
!kaggle datasets download -d paultimothymooney/chest-xray-pneumonia -p {data_dir}
````

## 4. Extract Archive

````python
zip_path = os.path.join(data_dir, 'chest-xray-pneumonia.zip')

print(f"Extracting {zip_path} ...")
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(data_dir)

print("Extraction complete.")
````

## 5. Clean Up and Reorganise

Remove nested folders and macOS metadata artefacts introduced by the zip archive,
then move the split folders to the expected location.

````python
unzip_path = os.path.join(data_dir, 'chest_xray')

# Remove macOS artefacts and nested duplicate folder
for artefact in ['chest_xray', '__MACOSX']:
    artefact_path = os.path.join(unzip_path, artefact)
    if os.path.exists(artefact_path):
        shutil.rmtree(artefact_path)
        print(f"Removed: {artefact_path}")
````

````python
# Move train, val, test splits to data/raw/
for folder in ['train', 'test', 'val']:
    src = os.path.join(unzip_path, folder)
    dst = os.path.join(data_dir, folder)
    shutil.move(src, dst)
    print(f"Moved: {folder} -> {dst}")

# Remove the now-empty extracted folder and the zip archive
shutil.rmtree(unzip_path)
os.remove(zip_path)

print("\nCleanup complete. Final directory structure:")
for split in ['train', 'val', 'test']:
    for cls in ['NORMAL', 'PNEUMONIA']:
        count = len(os.listdir(os.path.join(data_dir, split, cls)))
        print(f"  data/raw/{split}/{cls}: {count} images")
````
